# Pandas Basics
Pandas provides `Series` (1D) and `DataFrame` (2D) structures for handling tabular data efficiently.

## 1. Series

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], index=["a", "b", "c", "d"])
print(s)
print(s["b"])
print(s.values)
print(s.index)
```

## 2. DataFrame Creation

```python
data = {
    "Name": ["Yugal", "Aashma", "Rita", "Sam"],
    "Age": [25, 22, 28, 30],
    "Score": [85, 90, 78, 88]
}

df = pd.DataFrame(data)
df
```

## 3. Inspecting Data

```python
print(df.shape)
print(df.columns)
print(df.dtypes)
df.info()
df.describe()
```

## 4. Selecting Data (loc / iloc)

```python
print(df["Name"])            # single column
print("\n")
print(df[["Name", "Age"]])   # multiple columns
print("\n")
print(df.loc[0])              # row by label
print("\n")
print(df.iloc[1])             # row by position
print("\n")
print(df.loc[df["Age"] > 25]) # conditional filtering
```

## 5. Adding / Modifying Columns

```python
df["Passed"] = df["Score"] >= 80
df["Age_in_5_years"] = df["Age"] + 5
df
```

## 6. Sorting and Grouping

```python
print(df.sort_values("Score", ascending=False))

grouped = df.groupby("Passed")["Score"].mean()
print(grouped)
```

## 7. Handling Missing Data

```python
import numpy as np

df_missing = df.copy()
df_missing.loc[1, "Score"] = np.nan

print(df_missing.isnull().sum())
print(df_missing.fillna(df_missing["Score"].mean()))
print(df_missing.dropna())
```

## Summary
Pandas builds on NumPy to provide labeled, tabular data structures ideal for cleaning, exploring, and transforming datasets before feeding them into ML models.
