# Matplotlib & Seaborn Basics
Matplotlib is the core plotting library in Python; Seaborn is built on top of it for more attractive statistical visualizations with less code.

## 1. Matplotlib: Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)", color="blue")
plt.title("Simple Line Plot")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.show()
```

## 2. Matplotlib: Bar and Scatter Plots

```python
categories = ["A", "B", "C", "D"]
values = [23, 45, 12, 33]

plt.bar(categories, values, color="skyblue")
plt.title("Bar Chart")
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color="green")
plt.title("Scatter Plot")
plt.show()
```

## 3. Matplotlib: Histogram and Subplots

```python
data = np.random.randn(1000)

plt.hist(data, bins=30, color="orange", edgecolor="black")
plt.title("Histogram")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y, "o")
axes[0].set_title("Scatter (via subplot)")

axes[1].hist(data, bins=20, color="purple")
axes[1].set_title("Histogram (via subplot)")

plt.tight_layout()
plt.show()
```

## 4. Seaborn: Setup and Sample Data

```python
import seaborn as sns
import pandas as pd

tips = sns.load_dataset("tips")   # built-in sample dataset
tips.head()
```

## 5. Seaborn: Distribution and Categorical Plots

```python
sns.histplot(data=tips, x="total_bill", kde=True)
plt.title("Distribution of Total Bill")
plt.show()
```

```python
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Total Bill by Day")
plt.show()
```

## 6. Seaborn: Relationships Between Variables

```python
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex")
plt.title("Tip vs Total Bill")
plt.show()
```

```python
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

## Summary
- **Matplotlib**: low-level, highly customizable — good for full control over plots.
- **Seaborn**: high-level, statistical plots with better defaults and built-in dataset support.
- Both are commonly used together during EDA (Exploratory Data Analysis).
