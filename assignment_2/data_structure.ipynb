# Python Data Structures: List, Tuple, Set, Dictionary


## 1. List
- **Ordered**, **mutable**, allows **duplicates**
- Defined using square brackets `[]`

```python
# Creating a list
fruits = ["apple", "banana", "cherry", "apple"]
print(fruits)
print(type(fruits))
```

```python
# Indexing and slicing
print(fruits[0])      # first element
print(fruits[-1])     # last element
print(fruits[1:3])    # slicing
```

```python
# Common list methods
fruits.append("mango")      # add at end
fruits.insert(1, "kiwi")    # insert at index
fruits.remove("apple")      # remove first match
popped = fruits.pop()       # remove & return last item
fruits.sort()                # sort in place
print(fruits, "\n popped:", popped)
```

## 2. Tuple
- **Ordered**, **immutable**, allows **duplicates**
- Defined using parentheses `()`
- Faster than lists, used for fixed data

```python
# Creating a tuple
coordinates = (10, 20)
person = ("Yugal", 25, "Nepal")
print(coordinates)
print(person)
```

```python
# Indexing and unpacking
name, age, country = person
print(name, age, country)

# Tuples are immutable -> this would raise an error:
# coordinates[0] = 99
```

```python
# Tuple methods
nums = (1, 2, 2, 3, 4, 2)
print(nums.count(2))   # occurrences of 2
print(nums.index(3))   # index of first 3
```

## 3. Set
- **Unordered**, **mutable**, **no duplicates**
- Defined using curly braces `{}`
- Great for membership tests and removing duplicates

```python
# Creating a set
nums = {1, 2, 3, 3, 2}
print(nums)   # duplicates removed automatically
```

```python
# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Difference:", a ^ b)
```

```python
# Adding / removing elements
a.add(10)
a.discard(1)   # safely removes if present
print(a)
```

## 4. Dictionary
- Stores data as **key-value pairs**
- **Mutable**, keys must be unique
- Defined using curly braces `{key: value}`

```python
# Creating a dictionary
student = {
    "name": "Yugal",
    "age": 25,
    "course": "AI/ML Bootcamp"
}
print(student)
```

```python
# Accessing and updating values
print(student["name"])
student["age"] = 26          # update value
student["batch"] = 2026      # add new key
print(student)
```

```python
# Common dictionary methods
print(student.keys())
print(student.values())
print(student.items())

for key, value in student.items():
    print(f"{key}: {value}")
```

## Summary
| Structure | Ordered | Mutable | Duplicates | Syntax |
|---|---|---|---|---|
| List | Yes | Yes | Yes | `[]` |
| Tuple | Yes | No | Yes | `()` |
| Set | No | Yes | No | `{}` |
| Dictionary | Yes (insertion order) | Yes | Unique keys | `{key: value}` |
